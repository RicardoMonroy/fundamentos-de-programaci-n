# Seleccionador de alumnos

```python
import random
# Importamos la lista desde el archivo externo datos_alumnos.py
from datos_alumnos import alumnos

# Copiamos la lista para modificarla en memoria durante la clase
alumnos_pendientes = alumnos.copy()
```

```python
print(f"--- Selección de Alumnos para el Pizarrón ---")
print(f"Total de alumnos cargados: {len(alumnos_pendientes)}\n")


# .pop() selecciona y elimina el elemento en el índice indicado
indice_aleatorio = random.randrange(len(alumnos_pendientes))
seleccionado = alumnos_pendientes.pop(indice_aleatorio)

print(f"👉 Pasa al pizarrón: {seleccionado}")
print(f"Alumnos restantes por pasar: {len(alumnos_pendientes)}\n" + "-"*40)

```
