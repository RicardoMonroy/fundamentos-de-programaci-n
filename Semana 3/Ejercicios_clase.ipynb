```python
numero = float(input("Dame un número entero"))  

if numero % 2 == 0:
    print("Es número par")
else:
    print("Es número impar")
```

```python
suma = 0
while True:
    num = float(input("Ingresa un valor(Ingresa 0 para terminar): "))
    if num == 0:
        break
    suma += num
print(f"La suma de los valores es: {suma}")
```
