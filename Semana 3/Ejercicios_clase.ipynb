```python
#Ejercicio 6

dividendo = int(input("Dame un numero(mayor a el divisor): "))
divisor = int(input("Ingresa un divisor (No pongas 0): "))

if divisor <= 0:
    print("El número debe ser mayor a 0")
else:
    cociente = 0
    residuo = dividendo

    while residuo >= divisor:
        residuo = residuo - divisor
        cociente = cociente +1


print(f"{dividendo} / {divisor} = {cociente} con residuo {residuo}")
print(f"Termino el proceso")

```
