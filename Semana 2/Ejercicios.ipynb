# Pequeña ayuda

```python
import pdb

personas = int(input("Cuantas personas son?"))
edad_acumulada = 0

for p in range(1, personas + 1):
    print(f"Persona {p}")
    edad_persona = int(input(f"Dame la edad de la persona {p}"))
    edad_acumulada = edad_acumulada + edad_persona
    print(f"Su edad acumulada es {edad_acumulada}")

print("Fin del programa")
```

```python
acceso = True

while acceso:
    print("Ejemplo de un menú repetitivo")
    edad = int(input("Dame tu edad"))
    if edad < 18:
        print("Eres menor de edad, no tienes acceso")
    else:
        print("Acceso permitido")
        acceso = False

print("ya saliste del ciclo infinito")
```
