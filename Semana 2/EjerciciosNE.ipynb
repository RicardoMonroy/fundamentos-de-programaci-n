**Ejercicio 2:** Número par o impar
Enunciado: Crea un programa que pida un número entero y determine si es par o impar usando el operador módulo %. Entrada: 7 Salida: 7 es un número impar

```python
numero = int(input("Ingresa un número entero"))

if (numero % 2 == 0 ):
    print("Es un número par")
else:
    print("Es un número impar")
```

```python
tabla = int(input("Qué tabla quieres que muestre"))

# for(Int i = 1; i < 10; i++)
print("\nFor")
for m in range(1, 11):
    print(tabla, "x", m, "=", tabla * m)

print("\nWhile")
m = 1
while m <= 10:
    print(f"{tabla} x {m} = {tabla * m}")
    m += 1
```
